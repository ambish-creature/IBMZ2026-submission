```python
# Predict one post
model = joblib.load("classifier.joblib") 

text = ["In today’s rapidly evolving digital landscape, it is essential to leverage innovative solutions that foster meaningful engagement, drive sustainable growth, and unlock the full potential of emerging technologies"]

probabilities = model.predict_proba(text)[0]

ai_probability = probabilities[
    list(model.classes_).index(0)
]
human_probability = probabilities[
    list(model.classes_).index(1)
]

prediction = "AI-generated" if ai_probability >= 0.5 else "Human-written"

print(f"Prediction: {prediction}")
print(f"AI probability: {ai_probability:.1%}")
print(f"Human probability: {human_probability:.1%}")
```

```python
import subprocess
import sys
from pathlib import Path

project_dir = Path("fact_checker").resolve()

# Initialise database
subprocess.run(
    [sys.executable, "database.py"],
    cwd=project_dir,
    check=True
)

# Start FastAPI without blocking the notebook
server = subprocess.Popen(
    [
        sys.executable, "-m", "uvicorn",
        "app:app",
        "--host", "127.0.0.1",
        "--port", "8000"
    ],
    cwd=project_dir
)

print("Uvicorn started; PID:", server.pid)
```

```python
import requests

response = requests.post(
    "http://127.0.0.1:8000/fact-check",
    json={
        "text": "Donald Trump is the president of America",
        "language": "en"
    }
)

print(response.status_code)
data = response.json() 
result = data["results"][0]

print("Local database") 
print("Claim:", result["claim"])
print("Verdict:", result["verdict"])
print("Confidence:", result["confidence"])
print("Reasoning:", result["reasoning"])

sources = result["external_sources"]

print("External sources") 
for source in sources:
    print("Title:", source["title"])
    print("Snippet:", source["snippet"])
    print("URL:", source["url"])
    print()
```
